```python
x = df.drop('price',axis = 1)
y = df['price']
```

```python
x
```

```python
df.columns
```

```python
# onne hot encoding
x_one_encode = pd.get_dummies(x,drop_first=True,columns = ['model','transmission','fuelType'])
```

```python
x_one_encode
```

```python
x_one_encode = x_one_encode.astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()

columns = ['model','transmission','fuelType']
xlable = x.copy()
label_encoders = {}
```

```python
for col in columns:
    le = LabelEncoder()
    xlable[col] = le.fit_transform(xlable[col].astype(str))
    label_encoders[col] = le
```

```python
# use standard scaler to scale the data
from sklearn.preprocessing import StandardScaler
 
```

```python
numerical_cols = ['year','mileage','tax','mpg','engineSize']
scaler = StandardScaler()
x_one_encode[numerical_cols]= scaler.fit_transform(x_one_encode[numerical_cols])
```

```python
xlable.columns
```

```python
xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']] = scaler.fit_transform(xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg', 'engineSize']])
```

```python
xlable
```

```python
# model creation with one hot encoding
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_one_encode, y , test_size = 0.33,random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test,y_pred)
r2
```

```python
# adjustedr2
n = x_test.shape[0]
p = x_test.shape[1]
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
print('Adjusted r2:',adjusted_r2)
```
